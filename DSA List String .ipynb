```python
sales = [3,4,5,9,7]
```

```python
lar = sales[0]

for x in sales:
    if x > lar:
        lar = x
print(lar)
```

```python
numbers = [10,20,30,40]

total = 0

for x in numbers:
    total += x

average = total / len(numbers)

print(average)
```

```python
numbers = [10,15,13,34,56,23]


count = 0

for x in numbers:
    if x % 2 == 0:
        count += 1
print(count)
```

```python
odd_num = []

for x in numbers:
    if x % 2 != 0:
        odd_num.append(x)
print(odd_num)
```

```python
numbers = [10,15,13,34,56,23]

largest = numbers[0]
smallest = numbers[0]

for x in numbers:
    if x > largest:
        largest = x 
    if x < smallest:
        smallest = x

print(largest,smallest)
```

```python
# reverse a list
```

```python
numbers = [10,15,13,34,56,23]

reverse_list = []

for i in range(len(numbers)-1, -1, -1):
    reverse_list.append(numbers[i])
print(reverse_list)
```

```python
num = [10,-2,6,8,-5,-1,-6,0]
```

```python
pos = 0
neg = 0
zeros = 0

for x in num:
    if x > 0:
        pos += 1
    elif x < 0:
        neg += 1
    else:
        zeros += 1
print(pos,neg,zeros)
```

```python
sales = [1200,5000,3200,8000,2500]

high = sales[0]
low = sales[0]
total = 0


for x in sales:
    if x > high:
        high = x
    if x < low:
        low = x
    total += x
average = total / len(sales)

print(f" high sales {high}, lowest sales {low},total sales {total},average sales {average}")
```

```python
num = [10,45,22,90,55]

largest = num[0]

second_largest = float("-inf")

for x in num:
    if x > largest:
        second_largest = largest
        largest = x
    elif x > second_largest and x != largest:
        second_largest = x

print(second_largest)
```

```python
small = num[0]
sec_small = float("-inf")

for x in num:
    if x < small:
        sec_small = small
        small = x
    elif x < sec_small and x!= small:
        sec_small = x
print(sec_small)
```
